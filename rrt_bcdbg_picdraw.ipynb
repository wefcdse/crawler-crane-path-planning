```python
import matplotlib.pyplot as plt
import numpy as np

# ================= 1. 配置中文字体 =================
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'Arial Unicode MS'] 
plt.rcParams['axes.unicode_minus'] = False 

# ================= 2. 数据过滤与准备 =================
# 直接读取全局变量，不做任何防御性模拟设定
# 【核心逻辑】：只保留三个算法在该场景下都成功 (success_rate > 0.0) 的索引
valid_indices = [
    i for i in range(len(success_rate_list1))
    if success_rate_list1[i] > 0.0 and success_rate_list2[i] > 0.0 and success_rate_list3[i] > 0.0
]
print(f"✅ Filtering complete: total scenarios where all three algorithms succeeded: {len(valid_indices)} / {len(success_rate_list1)}")

# 提取对应成功场景下的最小动作切换次数
min_turns1 = [min_turn_count_list1[i] for i in valid_indices]
min_turns2 = [min_turn_count_list2[i] for i in valid_indices]
min_turns3 = [min_turn_count_list3[i] for i in valid_indices]

print(min_turns1)
print(min_turns2)
print(min_turns3)

# 定义分段边界 (与 avg_turn_count 保持一致): 1, 2, 3, 4, 7, 10, 15
bins = [-np.inf, 1 - 0.01, 2 - 0.01, 3 - 0.01, 4 - 0.01, 7 - 0.01, 10 - 0.01, 15 + 0.01, np.inf]
interval_labels = ['< 1', '1', '2', '3', '4 ~ 6', '7 ~ 9', '10 ~ 15', '> 15']

def count_in_bins(data, bins):
    counts = [0] * (len(bins) - 1)
    for val in data:
        for j in range(len(bins) - 1):
            if bins[j] <= val < bins[j+1]:
                counts[j] += 1
                break
    return counts

counts1 = count_in_bins(min_turns1, bins)
counts2 = count_in_bins(min_turns2, bins)
counts3 = count_in_bins(min_turns3, bins)

def to_percent(counts):
    total = sum(counts)
    if total == 0: return [0]*len(counts)
    return [c / total * 100 for c in counts]

p1 = to_percent(counts1)
p2 = to_percent(counts2)
p3 = to_percent(counts3)


# ================= 3. 绘制堆叠柱状图 =================
fig, ax = plt.subplots(figsize=(10, 6.5), dpi=300)

labels_bar = ['RRT-BC\n(Proposed)', 'Improved RRT\n(Baseline)', 'Uniform RRT']
x = np.arange(len(labels_bar))
width = 0.5

# 8种颜色：从蓝(极少切换/极平滑)渐变到红(频繁切换/繁琐)
colors = ['#4575b4', '#74add1', '#abd9e9', '#e0f3f8', '#fee090', '#fdae61', '#f46d43', '#d73027'] 

bottom = np.zeros(3)
percentages_matrix = [p1, p2, p3]

for i in range(8):
    vals = [percentages_matrix[0][i], percentages_matrix[1][i], percentages_matrix[2][i]]
    # 绘制堆叠柱
    ax.bar(x, vals, width, bottom=bottom, label=interval_labels[i], 
           color=colors[i], edgecolor='black', linewidth=1.0, zorder=3)
    
    # 在色块内部添加百分比文本
    for j, val in enumerate(vals):
        if i in [0, 1, 6, 7]: # 深色背景(深蓝、深红)
            text_color = 'white'
        else: # 浅色背景(浅蓝、黄、浅橙)
            text_color = 'black'
        if val > 5.0:  # 占比大于5%时显示标签
            # 自动判断背景深浅以切换文字颜色
      
            ax.text(x[j], bottom[j] + val / 2, f'{val:.1f}%', 
                    ha='center', va='center', fontsize=11, fontweight='bold', color=text_color)
        elif val > 2.5: # 占比极小但大于0.02
            ax.text(x[j], bottom[j] + val / 2, f'{val:.1f}%', 
                    ha='center', va='center', fontsize=9, fontweight='bold', color=text_color)
    bottom += vals

# ================= 4. 图表修饰 =================
ax.set_ylabel('Proportion (%)', fontsize=14, fontweight='bold')
ax.set_title('Comparison of the minimum number of action switches of the optimal paths in successfully planned scenarios', fontsize=16, fontweight='bold', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(labels_bar, fontsize=13)
ax.set_ylim(0, 105)
ax.set_yticks(np.arange(0, 101, 20))

# 【核心修改】图例放在底部，ncol=4 排成两排，mode="expand" 保证宽度与图表一致
ax.legend(loc='upper center', 
          bbox_to_anchor=(0.05, -0.2, 0.9, 0.1),  # 稍微往下移一点(-0.32)给两排图例留空间
          ncol=4,                             # 8个图例分4列，正好排成 4+4 的两排
          mode="expand",                      # 强制拉伸至指定宽度
          borderaxespad=0, 
          frameon=False,                      # 去掉边框
          fontsize=11,
          columnspacing=1.5,
        #   title='最小动作切换次数区间', 
          title_fontsize=12)

# 极简学术风格边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.yaxis.grid(True, linestyle='--', alpha=0.6, zorder=0)
ax.set_axisbelow(True) 


# ================= 5. 保存与显示 =================

# 自动调整布局以包含外部图例
plt.tight_layout()

plt.savefig('fig_min_turn_count_stacked_bar.pdf', bbox_inches='tight')
plt.savefig('fig_min_turn_count_stacked_bar.png', bbox_inches='tight')
plt.savefig('fig_min_turn_count_stacked_bar.svg', bbox_inches='tight', format='svg')

print("✅ Stacked bar chart of the minimum number of action switches saved as: fig_min_turn_count_stacked_bar.pdf and .png")
plt.show()
```
